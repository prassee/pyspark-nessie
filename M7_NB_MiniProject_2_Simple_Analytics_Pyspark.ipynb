#### Normalization or standardization

select the applicable features using vector assembler and apply scaling using `MinMaxScaler` from pyspark

Hint: [MinMaxScaler](https://spark.apache.org/docs/latest/ml-features#minmaxscaler)

```python
# YOUR CODE HERE
from pyspark.ml.feature import VectorAssembler
from pyspark.sql import SparkSession
from pyspark.ml.feature import MinMaxScaler

houses_raw = matches
# Select feature columns for scaling (excluding 'No' as it's just an index and target variable)
feature_cols = [
    "X1 transaction date",
    "X2 house age",
    "X3 distance to the nearest MRT station",
    "X4 number of convenience stores",
    "X5 latitude",
    "X6 longitude",
]

# Create vector assembler for features
assembler = VectorAssembler(inputCols=feature_cols, outputCol="features_raw")
houses_assembled = assembler.transform(houses_raw)

# Apply MinMaxScaler
scaler = MinMaxScaler(inputCol="features_raw", outputCol="features_scaled")
scaler_model = scaler.fit(houses_assembled)
houses_scaled = scaler_model.transform(houses_assembled)

# Show the scaled features
houses_scaled.select("features_scaled", "Y house price of unit area").show(
    5, truncate=False
)
```

#### Feature selection and splitting the data

```python
# YOUR CODE HERE
houses_scaled = scaler_model.transform(houses_assembled)
# Split the data into training and testing sets
train_data, test_data = houses_scaled.randomSplit([0.8, 0.2], seed=42)

print(f"Training set size: {train_data.count()}")
print(f"Test set size: {test_data.count()}")
```

### Train and Evaluate the model (2 points)

Using LinearRegression from `pyspark.ml` fit the data and find the coefficients and intercept

```python
regressor = LinearRegression(
    featuresCol="features_scaled", labelCol="Y house price of unit area"
)

# Fit the model on training data
lr_model = regressor.fit(train_data)

# Get coefficients and intercept
print("Coefficients:", lr_model.coefficients)
print("Intercept:", lr_model.intercept)

# Print feature names for better understanding
feature_names = [
    "X1 transaction date",
    "X2 house age",
    "X3 distance to the nearest MRT station",
    "X4 number of convenience stores",
    "X5 latitude",
    "X6 longitude",
]

print("\nFeature Coefficients:")
for i, (feature, coeff) in enumerate(zip(feature_names, lr_model.coefficients)):
    print(f"{feature}: {coeff}")
```
